```python
from models import *
from utils import *
num_classes = 10
dataset ="cifar10"
```

```python
def check_missing_layers(model):
    state_dict_keys = set(model.state_dict().keys())
    named_param_keys = set(name for name, _ in model.named_parameters())
    
    # Find the difference
    missing_in_named_parameters = state_dict_keys - named_param_keys
    missing_in_state_dict = named_param_keys - state_dict_keys
    
    return missing_in_named_parameters, missing_in_state_dict


```

```python
model_name = "shufflenet"

model_fn = partial(models.get_model(model_name)[
                        0], num_classes=num_classes, dataset=dataset)

model0 = model_fn().to(device)
# Example usage:
missing_in_named_parameters, missing_in_state_dict = check_missing_layers(model0)

print("Missing in named_parameters:", missing_in_named_parameters)
print("Missing in state_dict:", missing_in_state_dict)

# created models 
model0 = model_fn().to(device) 
```
