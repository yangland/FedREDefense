```python
import hdbscan
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# Parameters
n_samples = [30,70]
n_clusters = 2
centers = np.array([[0, 0], [1, 1]])  # Centers of the clusters
cluster_std = [1, 1]  # Different variances for clusters

# Generate synthetic data
X, y = make_blobs(n_samples=n_samples, centers=centers, cluster_std=cluster_std, random_state=42)

# Plot
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', alpha=0.6, edgecolors='k')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Synthetic 2D Gaussian Clusters")
plt.show()
```

```python
clusterer = hdbscan.HDBSCAN(min_cluster_size=sum([30,70])//2 + 1,min_samples=1, allow_single_cluster=True).fit(X)
```

```python


# Generate synthetic data
X, y = make_blobs(n_samples=n_samples, centers=centers, cluster_std=cluster_std, random_state=42)

# Plot
highlight_indices = np.where(clusterer.labels_ == 0)[0]  # Indices of cluster 1

plt.scatter(X[:, 0], X[:, 1], c='lightgray', alpha=0.6, edgecolors='k')  # Base plot
plt.scatter(X[highlight_indices, 0], X[highlight_indices, 1], c='red', edgecolors='k', label="Highlighted")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Highlighted Synthetic Clusters")
plt.legend()
plt.show()
```
