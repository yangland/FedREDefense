```python
from collections import Counter

def replaced_params_count(replaced_params_list):
    """
    Counts occurrences of values in the 'values' lists across multiple replaced_params dictionaries.

    :param replaced_params_list: List of replaced_params dictionaries.
    :return: Counter with the count of each value.
    """
    count = Counter()
    for replaced_params in replaced_params_list:
        for values in replaced_params.values():
            count.update(values)
    return count

# Example usage:
replaced_params_list = [
    {"layer1": [1, 2, 3], "layer2": [2, 3, 4]},
    {"layer1": [3, 4, 5], "layer3": [1, 5, 6]}
]

result = replaced_params_count(replaced_params_list)
print(result)  # Output: Counter({3: 3, 2: 2, 4: 2, 1: 2, 5: 2, 6: 1})
```

```python
from collections import Counter
import math

def top_n_percent(counter, percent=10):
    """
    Returns the top N% of elements from a Counter.
    
    :param counter: Counter object with value counts.
    :param percent: Percentage of elements to keep (default 10%).
    :return: Dictionary with the top N% elements.
    """
    if not counter:
        return {}

    sorted_items = counter.most_common()  # Sort by count (descending)
    top_n = math.ceil(len(sorted_items) * (percent / 100))  # Compute top N elements
    return dict(sorted_items[:top_n])

# Example usage:
counter = Counter({3: 2, 1: 1, 2: 1, 4: 1, 5: 1})
result = top_n_percent(counter, 10)  # Get top 10%

print(result)  
```

```python
import torch
import matplotlib.pyplot as plt
import seaborn as sns

def plot_layer_weights(layer, layer_name, save_plot=False):
    """
    Plots the weights of a specific layer in a PyTorch model as a heatmap.
    
    Parameters:
        layer (torch.nn.Module): The PyTorch layer (e.g., torch.nn.Linear, torch.nn.Conv2d).
        layer_name (str): A text string to label the plot and use in the filename.
        save_plot (bool): If True, saves the plot as a PNG file with the layer_name as part of the filename.
    """
    # Check if the layer has weights
    if not hasattr(layer, 'weight'):
        raise ValueError(f"The provided layer does not have weights.")
    
    # Extract the weights
    weights = layer.weight.data.cpu().numpy()
    
    # Plot the weights as a heatmap
    plt.figure(figsize=(10, 6))
    sns.heatmap(weights, cmap="coolwarm", annot=False, cbar=True)
    plt.title(f'Weight Matrix of {layer_name}')
    plt.xlabel('Neuron Index')
    plt.ylabel('Input Index')
    
    if save_plot:
        # Save the plot with the layer_name as part of the filename
        filename = f"{layer_name}_weights_heatmap.png"
        plt.savefig(filename, bbox_inches='tight', dpi=300)
        print(f"Plot saved as {filename}")
    
    plt.show()

# Example usage:
model = torch.nn.Sequential(
    torch.nn.Linear(10, 5),
    torch.nn.ReLU(),
    torch.nn.Linear(5, 2)
)

# Access the second Linear layer directly
second_linear_layer = model[2]
plot_layer_weights(second_linear_layer, 'Second_Linear_Layer', save_plot=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Function to generate a sphere
def generate_sphere(radius, center, resolution=100):
    u = np.linspace(0, 2 * np.pi, resolution)
    v = np.linspace(0, np.pi, resolution)
    x = center[0] + radius * np.outer(np.cos(u), np.sin(v))
    y = center[1] + radius * np.outer(np.sin(u), np.sin(v))
    z = center[2] + radius * np.outer(np.ones(np.size(u)), np.cos(v))
    return x, y, z

# Parameters
r = 1  # Radius of the sphere (example, you can replace with ||Δ_B||)
center = [0, 0, 0]  # Center of the sphere
U = np.pi / 14  # Angle of the cone (example, replace with your U)
delta_B = np.array([1, 0, 0])  # Vector for W_B (example, replace with your W_B)
W_B = delta_B / np.linalg.norm(delta_B)  # Normalize the vector W_B

# Generate sphere
x, y, z = generate_sphere(r, center)

# Create the plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the sphere
ax.plot_surface(x, y, z, color='b', alpha=0.3)

# Plot the blue vector (W_B) from the center to the surface
ax.quiver(center[0], center[1], center[2], W_B[0], W_B[1], W_B[2], length=r, color='b', linewidth=2)

# Cone parameters
cone_angle = 2 * U  # Full angle of the cone
cone_height = r  # Height of the cone from the center to the surface of the sphere

# Define a few points on the cone's surface using polar coordinates
theta = np.linspace(0, 2 * np.pi, 30)  # Reduced points for performance
phi = np.linspace(np.pi / 2 - cone_angle / 2, np.pi / 2 + cone_angle / 2, 30)  # Reduced points

# Generate cone surface (adjusted to start at center and intersect the sphere)
theta_grid, phi_grid = np.meshgrid(theta, phi)
x_cone = cone_height * np.outer(np.cos(theta_grid), np.sin(phi_grid)) + center[0]
y_cone = cone_height * np.outer(np.sin(theta_grid), np.sin(phi_grid)) + center[1]
z_cone = cone_height * np.outer(np.ones(np.size(theta_grid)), np.cos(phi_grid)) + center[2]

# Plot the cone surface (lower transparency to make it more visible)
ax.plot_surface(x_cone, y_cone, z_cone, color='g', alpha=0.1)

# Red vector from center to edge of the cone (intersection with the sphere)
# Calculate the point where the cone intersects the sphere
mali = center + W_B * r * np.cos(U)  # Point where the cone intersects the sphere

# Plot the red vector from center to the edge of the cone
ax.quiver(center[0], center[1], center[2], mali[0] - center[0], mali[1] - center[1], mali[2] - center[2], color='r', linewidth=2)

# Set labels and aspect
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title("3D Sphere, Vector and Cone Visualization")

# Set the aspect ratio
ax.set_box_aspect([1, 1, 1])  # Equal scaling in all directions

# Show the plot
plt.show()
```
